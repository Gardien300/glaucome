# Évaluation du modèle (dossier 0) sur le dossier 1

Charge le modèle et le scaler entraînés sur le dossier 0, extrait les features du **dossier 1**, et affiche les métriques.  
À exécuter après avoir lancé le notebook **03_training_ml** et la cellule de sauvegarde (section 6).

```python
import sys
from pathlib import Path
PROJECT_ROOT = Path.cwd() if (Path.cwd() / "train_labels.csv").exists() else Path.cwd().parent
sys.path.insert(0, str(PROJECT_ROOT))

import joblib
from src.data_loading import get_subset_for_folder
from src.feature_extraction import build_feature_matrix
from src.evaluation_metrics import print_metrics
```

## 1. Charger le modèle et le scaler

```python
models_dir = PROJECT_ROOT / "models"
clf = joblib.load(models_dir / "rf_clf.joblib")
scaler = joblib.load(models_dir / "scaler.joblib")
print("Modèle et scaler chargés.")
```

## 2. Charger les données du dossier 1

```python
df1 = get_subset_for_folder("1")
print("Dossier 1:", len(df1), "images")
print(df1["class"].value_counts())
```

## 3. Extraction des features et prédiction

On peut limiter le nombre d'échantillons (max_samples) pour un test rapide, ou mettre None pour tout le dossier.

```python
MAX_SAMPLES = 2000  # Mettre None pour tout le dossier 1

X1, y1 = build_feature_matrix(df1, root=PROJECT_ROOT, feature_type="simple", max_samples=MAX_SAMPLES)
X1_s = scaler.transform(X1)
y1_pred = clf.predict(X1_s)
y1_score = clf.predict_proba(X1_s)[:, 1]
print("Shapes:", X1.shape)
```

## 4. Métriques sur le dossier 1

```python
print("=== Métriques sur le DOSSIER 1 (modèle entraîné sur dossier 0) ===")
print_metrics(y1, y1_pred, y1_score)
```
